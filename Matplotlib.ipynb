```python
# Plotting the first gragh
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[7,8,9,10,11,12]
plt.plot(x,y)
plt.show()
```

```python
# Plotting a line using array elements
import matplotlib.pyplot as plt
import numpy as np
# X-Axis points
x=np.array([12,34,56,67])
# Y-Axis points
y=np.array([89,67,90,34])
plt.plot(x,y)
plt.show()
```

```python
# Plotting without a line
import matplotlib.pyplot as plt
import numpy as np
# x points
xpoints=np.array([2,4,6,8])
ypoints=np.array([9,5,2,6])
plt.plot(xpoints,ypoints,'o')
plt.grid()
plt.show()
```

```python
# Plotting without the x axis points
import matplotlib.pyplot as plt
import numpy as np

y=np.array([4,6,7,8,9,10])
plt.plot(y)
plt.show()
```

```python
# ADDING A TITLE IN PLOT
import matplotlib.pyplot as plt

months=['Jan','Feb','Mar','Apr','May']
sales=[100,150,120,180,200]

plt.plot(months,sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.grid()
plt.show()
```

```python
# ADDING MARKERS IN THE PLOT GRAGH
import matplotlib.pyplot as plt

months=['Jan','Feb','Mar','Apr','May']
sales=[100,150,120,180,200]

plt.plot(months,sales,'o-r', ms=20,mec='black',mfc='yellow')

plt.title('Monthly Sales ')
plt.xlabel('Months')
plt.ylabel('Sales')
# Saving the visual
plt.savefig('Monthly_Sales.png')

plt.grid()
plt.show()
```

```python
# MATPLOTLIB FIRST PROGRAM
import matplotlib.pyplot as plt
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
study_hours = [2, 4, 3, 5, 6]

plt.plot(days,study_hours)
plt.title('Study Hours Per Day')
plt.xlabel('Day')
plt.ylabel('Study Hours')
plt.grid()
plt.show()
```
